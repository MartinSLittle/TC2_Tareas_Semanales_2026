### Introducción

En este trabajo, se realizará el análisis de las transferencia de órdenes dos, tres y cuatro de un filtro pasa-bajos mediante la aproximación de stortch, basándose en los polinomios de Bessel-Thompson. ### Polinomios de Bessel-Thompson: Aproximación de Stortch

La aproximación de Stortch es un método que se utiliza en conjunto con los polinomios de Bessel-Thompson. Todo este análisis se realiza bajo la premisa de un retardo de fase unitario y, además, de la condición de máxima planicidad de la fase. Esto quiere decir que la función retardo, la cuál se puede expresar como la tasa de variación instantanea negativa de la función fase con respecto a omega, debe ser cero.

Dicho esto, buscamos primero la transferencia de orden n = 2:

$$
\begin{array}{l}
\text{para } n=2. \\[10pt]
\text{sabiendo que :} \\[10pt]
\text{cotgh}(s) = \frac{\cosh(s)}{\sinh(s)} = \frac{1}{s} + \frac{1}{\frac{3}{s} + \frac{1}{\frac{5}{s} + \frac{1}{\text{etc.}}}} \\[20pt]
\text{para } n=2\text{, lo trunco en el} \\
\text{primer Bucle.} \\[10pt]
\text{cotgh}(s) = \frac{1}{s} + \frac{1}{\frac{3}{s} + \underbrace{0}_{\text{truncado}}} = \frac{s^2 + 3}{3 \cdot s} \longrightarrow 
\left\{
\begin{array}{l}
s^2 + 3 = \cosh(s) \\
3 \cdot s = \sinh(s)
\end{array}
\right. \\[25pt]
\text{La expresión genérica} \\
\text{indica que :} \\[10pt]
T_n(s) = \frac{B_n(0)}{B_n(s)} \\[15pt]
\text{donde, según enunciado } \longrightarrow D(\omega=0) = 1 seg \qquad \left( \begin{array}{l} \text{condición para usar} \\ \text{Bessel tal cual} \end{array} \right) \\[15pt]
T_2(s) = \frac{B_2(0)}{B_2(s)} = \frac{B_2(0)}{\cosh(s) + \sinh(s)} = \frac{3}{s^2 + 3s + 3}
\end{array}
$$

De la misma manera, realizamos la aproximación para n = 3:

$$
\begin{array}{l}
\text{Para } n=3: \\[10pt]
T_3(s) . \text{ Realizamos el mismo procedimiento, pero truncando luego de dos bucles.} \\[10pt]
\text{cotgh}(s) = \frac{1}{s} + \frac{1}{\frac{3}{s} + \frac{1}{\frac{5}{s} + 0}} = \frac{5s}{s^2 + 15} \\[25pt]
T_3(s) = \frac{B_3(0)}{B_3(s)} = \frac{15}{s^3 + 6 \cdot s^2 + 15 \cdot s + 15} \\[40pt]
T_4(s) . \text{Nuevamente continuamos con el mismo procedimiento, pero }\\[10pt]
\text{cotgh}(s) = \frac{1}{s} + \frac{1}{\frac{3}{s} + \frac{1}{\frac{5}{s} + \frac{1}{\frac{7}{s} + 0}}} \\[35pt]
T_4(s) = \frac{105}{s^4 + 10 \cdot s^3 + 45 \cdot s^2 + 105 \cdot s + 105}
\end{array}
$$

De esta manera, obtenemos lo solicitado para el ejercicio 1.

### Retardo de grupo no unitario

A pesar de haber obtenido la función T($) que cumple con las condiciones que queremos, ahora se obtendrá la función retardo D(w) y se analizará un retardo no unitario. Para ello, utilizaremos las siguientes ecuaciones:

$$
\left\{
\begin{array}{l}
D(\omega) = \frac{-d\theta(\omega)}{d\omega} \\[15pt]
\theta(\omega) = -\arctan\left( \frac{\text{Img}(B_n(j\omega))}{\text{Re}(B_n(j\omega))} \right)
\end{array}
\right.
$$

Si trabajamos un poco algebraicamente, llegamos a la siguiente expresión. Para la expresión final y evitar realizar derivadas y cometer un error algebraico, obtenemos del Schaumann la expresión final:

$$
\begin{array}{l}
T_3(s): \\[10pt]
\qquad B_3(s) = s^3 + 6s^2 + 15 \cdot s + 15 \\[10pt]
\qquad B_3(j\omega) = -j \cdot \omega^3 + 15\omega j - 6 \cdot \omega^2 + 15 \\[15pt]
\theta(\omega) = -\arctan\left( \frac{-\omega^3 + 15\omega}{-6\omega^2 + 15} \right) \\[20pt]
\text{En el schawmann la expresión} \\
\text{luego de trabajar la derivada es:} \\[15pt]
\qquad D_3 = \frac{6\omega^4 + 45\omega^2 + 225}{\omega^6 + 6\omega^4 + 45\omega^2 + 225} \\[20pt]
D_3(\omega = 2,5) = 0,7520 \\[15pt]
\% \text{ err :} \qquad \quad 1 \longrightarrow 100\% \\
\qquad \qquad \quad 0,752 \longrightarrow x\% \\[10pt]
\qquad \qquad \quad x = 75,2\% \\[15pt]
100\% - 75,2\% = 24,8\% \\[10pt]
\% \text{ err} = 24,8\% //

$$

Esto quiere decir que el error de desviación por contemplar una fase no unitaria es del 24,8%

### Sintetización de red circuital Sallen-Key n = 3

Finalmente, para obtener una sintetización de la transferencia con la que se viene trabajando, como la red circuital Sallen-Key es una red circuital de orden n = 2, factorizamos el denominador de la transferencia T($) de orden 3 y utilizamos en cascada dos filtros pasa-bajos que cumplan con los requisitos que se necesitan.

Para el desarrollo del filtro Sallen-Key:

$$
\begin{array}{l}
\text{Estructura Sallen-Key , } K=1. \\[15pt]
\text{Método de nodos :} \\[10pt]
\left\{
\begin{array}{ll}
(G_1 + G_2 + sC_2) \cdot V_a - V_b \cdot G_2 - V_i \cdot G_1 - V_o \cdot sC_2 = 0 & \\[5pt]
(G_2 + sC_1) \cdot V_b - V_a \cdot G_2 = 0 & \\[5pt]
V_o = V_b &
\end{array}
\right. \\[25pt]
V_a = \frac{(G_2 + sC_1)}{G_2} \cdot V_o \\[20pt]
(G_1 + G_2 + sC_2) \cdot \underbrace{\frac{(G_2 + sC_1)}{G_2}}_{1 + sC_1 \cdot R_2} \cdot V_o - V_o \cdot G_2 - V_i \cdot G_1 - V_o \cdot sC_2 = 0 \\[10pt]
\hspace{7cm} -V_o \cdot (G_2 + sC_2) = V_i \cdot G_1 \\[15pt]
(G_1 + \cancel{G_2} + \cancel{sC_2} + G_1 \cdot R_2 \cdot sC_1 + sC_1 + s^2 \cdot C_1 \cdot C_2 \cdot R_2 - \cancel{G_2} - \cancel{sC_2}) \cdot V_o = V_i \cdot G_1 \\[20pt]
\frac{V_o}{V_i} = \frac{G_1}{s^2 \cdot C_1 \cdot C_2 \cdot R_2 + s \cdot (C_1 \cdot \frac{R_2}{R_1} + C_1) + G_1} \\[25pt]
\hspace{0.5cm} = \frac{G_1 \cdot G_2 / C_1 \cdot C_2}{s^2 + \underbrace{s \cdot \left( \frac{\cancel{C_1} \cdot \frac{R_2}{R_1} + \cancel{C_1}}{\cancel{C_1} \cdot C_2 \cdot R_2} \right)}_{\displaystyle s \cdot \left( \frac{1}{R_1 \cdot C_2} + \frac{1}{R_2 \cdot C_2} \right)} + 1 / R_1 \cdot R_2 \cdot C_1 \cdot C_2} \\[35pt]
\hspace{4cm} \frac{R_2 \cdot \cancel{C_2} + R_1 \cdot \cancel{C_2}}{R_1 \cdot R_2 \cdot C_2^{\cancel{2}}} \\[15pt]
\hspace{4cm} \frac{R_1 + R_2}{R_1 \cdot R_2 \cdot C_2} \\[25pt]
T(s) = \frac{V_o}{V_i} = \frac{1 / R_1 \cdot R_2 \cdot C_1 \cdot C_2}{s^2 + s \cdot \left( \frac{1}{R_1 \cdot C_2} + \frac{1}{R_2 \cdot C_2} \right) + 1 / R_1 \cdot R_2 \cdot C_1 \cdot C_2}
\end{array}
$$

Teniendo la transferencia del Sallen-Key con ganancia K = 1, procedemos a analizar la transferencia de orden n = 3 y factorizar para cumplir con las condiciones impuestas según los valores de la transferencia:

$$
\begin{array}{l}
T_3(s) = \frac{15}{s^3 + 6 \cdot s^2 + 15 \cdot s + 15} \quad \xrightarrow{\text{factorizando}} \\[15pt]
T_3(s) = \frac{15}{(s + 2,3222) \cdot (s^2 + 3,6778s + 6,4594)} \\[30pt]
T_3(s) = 15 \cdot \underbrace{\frac{1}{s + 2,3222}}_{T_{-1}(s)} \cdot \underbrace{\frac{1}{s^2 + 3,6778 \cdot s + 6,4594}}_{\begin{array}{c} T_{-2}(s) \\ \text{Sallen - Key} \end{array}} \quad \xrightarrow{\begin{array}{c} \text{Para poder decir} \\ \text{que } T_1 \cdot T_2 \\ \downarrow \\ K_1 \cdot K_2 = 1 \end{array}} \\[35pt]
15 = 2,3222 \cdot 6,4594 \\[15pt]
T_{-1}(s) = \frac{2,3222}{s + 2,3222} \\[15pt]
T_{-2}(s) = \frac{6,4594}{s^2 + 3,6778 \cdot s + 6,4594} \\[25pt]
\text{Etapa Sallen - Key :} \\[10pt]
\omega_o^2 \longrightarrow \omega_o = 2,541535 \qquad \frac{\omega_o}{Q} = 3,6778 = \frac{1}{R_1 \cdot C_2} + \frac{1}{R_2 \cdot C_2} \\[15pt]
2,541535 = \frac{1}{R_1 \cdot R_2 \cdot C_1 \cdot C_2} \qquad Q = 0,691 \swarrow \\[15pt]
\Omega_z = R_1 = R_2 = 1 \Omega \qquad \qquad 3,6778 = \frac{2}{C_2} \longrightarrow C_2 = 0,5438 \text{ F} \\[15pt]

C_1 = \frac{1}{2,541535 \cdot C_2} = 0,7235436 \text{ F} \\[20pt]

\left\{

\right. \\[35pt]
